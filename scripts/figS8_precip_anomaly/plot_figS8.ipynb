```python
# 读数据
data_path = str(DATA / "anoaprc.mat")
data = sio.loadmat(data_path)
lon2d = sio.loadmat(data_path)['lon2d']
lat2d = sio.loadmat(data_path)['lat2d']
ano_summer = data['ano_summer'].T
ano_winter = data['ano_winter'].T
ano_annual = data['ano_annual'].T
hannual = data['hannual'].T
hsummer = data['hsummer'].T
hwinter = data['hwinter'].T
```
